```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load CSV file (skip first row as it's a header)
file_name = "Bunch_crossing_2017.csv"  # Replace with your actual file name

# Read CSV, skipping the first row (header row)
df = pd.read_csv(file_name, skiprows=1, names=["Fill", "Run", "Bx"])

# Convert Bx column to numeric values
df["Bx"] = pd.to_numeric(df["Bx"], errors='coerce')

```

```python
df["Bx"]
```

```python
# Plot histogram
bins = np.linspace(0, 2600, 27)  # 21 edges for 20 bins
plt.figure(figsize=(8, 5))
plt.hist(df["Bx"], bins=bins, color='blue', alpha=0.7, edgecolor='black')

# Formatting
plt.xlabel("Number of colliding bunches per fill", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.title("", fontsize=14)
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Save and show plot
plt.savefig("bx_histogram_2017.pdf")
plt.show()
```

```python
# Make a plot with 3 lists
events = [13883679, 16791494, 23300931]
skipped = [655064, 1577918, 1274815]
leftover_arr = []
for event, skip in (zip(events, skipped)):
    leftover = (skip/event)*100
    leftover_arr.append(leftover)
    print("leftover ", leftover)
```

```python
plt.figure(figsize=(6, 4))
plt.bar(range(len(leftover_arr)), leftover_arr, color='blue', alpha=0.7, edgecolor='black')

# Formatting
plt.xticks(range(len(leftover_arr)), [f"Dataset {i}" for i in range("2016", "2017")])
plt.xlabel("Datasets")
plt.ylabel("Leftover (%)")
plt.title("Leftover Percentage per Dataset")
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Save and show plot
plt.savefig("leftover_histogram.pdf")
plt.show()
```
