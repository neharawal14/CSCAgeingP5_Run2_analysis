```python
import pandas as pd
import datetime
year = "2018"
text_file = "stable_beam_time_"+year+".csv"

# Read the fill information from the text file into a DataFrame
df = pd.read_csv(text_file, sep=',', engine='python')
```

```python
df = df.dropna()
```

```python
df['start time'] = pd.to_datetime(df['start time'], format = '%Y-%m-%d %H:%M:%S') 
print(df['start time'])
# Add 30 minutes to create 'half hour end' column
df['half hour end'] = df['start time'] + pd.Timedelta(minutes=30)
```

```python

# Convert both columns to epoch time (Unix timestamp in seconds)
df['start time (epoch)'] = df['start time'].astype('int64') // 10**9
df['half hour end (epoch)'] = df['half hour end'].astype('int64') // 10**9

# Save only the epoch timestamps to a new CSV file
output_file = "epoch_stable_beam_time_"+year+".csv"
df[['start time (epoch)', 'half hour end (epoch)']].to_csv(output_file, index=False)

print(f"Updated CSV file saved as: {output_file}")
```
